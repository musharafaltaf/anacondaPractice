# Matplotlib
<pre><b>it is data visualisation technique.
Matplotlib is a comprehensive pyhton libaray for creating static,animated and interactive vusualisations.</pre>

```python
import matplotlib.pyplot as plt
```

```python
#line plot
x = [2,6,8,4,9,3,4,5,6,7,8,8]
y = [1,3,5,7,6,4,3,3,2,2,1,7]
plt.plot(x,y)
plt.show()
```

```python
x = [2,6,8,4,9,3,4,5,6,7,8,8]
y = [1,3,5,7,6,4,3,3,2,2,1,7]
plt.plot(x,y)
plt.title("line plot")
plt.xlabel('x-Axis')
plt.ylabel('y-Axis')
plt.grid()
plt.show()
```

```python
x = [2,6,8,4,9,3,4,5,6,7,8,8]
y = [1,3,5,7,6,4,3,3,2,2,1,7]
plt.plot(x,y,color='r',)
plt.title("line plot",color='b')
plt.xlabel('x-Axis',color='orange')
plt.ylabel('y-Axis',color='yellow')
plt.grid(color='m')
plt.show()
```

```python
x = [2,6,8,4,9,3,4,5,6,7,8,8]
y = [1,3,5,7,6,4,3,3,2,2,1,7]
plt.plot(x,y,'og--')  # 'o' for dot and 'g' for color
plt.title("line plot",color='b')
plt.xlabel('x-Axis',color='orange')
plt.ylabel('y-Axis',color='yellow')
plt.grid(color='m')
plt.show()
```

```python
#Bar plot
x = [1,3,5,7,9]
y = [2,4,6,8,10]
plt.bar(x,y,color=['b','g','r'])
plt.title("bar plot")
plt.xlabel('x-Axis')
plt.ylabel('y-Axis')
plt.annotate("haroon",xy=(4.6,6.5))
plt.show()
```

```python
x = [1,3,5,7,9]
y = [2,4,6,8,10]
plt.bar(x,y,width=1,label='Example1')
x1 = [4,7,2,9,8]
y2 = [9,2,6,4,1]
plt.bar(x1,y2,label='Example2')
plt.title("bar plot")
plt.legend()
plt.xlabel('x-Axis')
plt.ylabel('y-Axis')
plt.annotate("haroon",xy=(4.6,6.5))
plt.show()
```

```python
x = [1,3,5,7,9]
y = [2,4,6,8,10]
plt.bar(x,y,width=1,label='Example1')
x1 = [4,7,2,9,8]
y2 = [9,2,6,4,1]
plt.bar(x1,y2,label='Example2')
plt.show()
plt.title("bar plot")
plt.legend()
plt.xlabel('x-Axis')
plt.ylabel('y-Axis')
plt.annotate("haroon",xy=(4.6,6.5))
plt.show()
```

```python
#Histograms
age = [10,8,5,20,3,15,13,1,25,35,28,48,40,45]
bins = [0,10,20,30,40,50]
plt.hist(age,bins,rwidth=0.2,edgecolor='k',linewidth=2,color='orange')
plt.title("Histogram")
plt.show()
```

```python
#scatter plot
x = [3,4,6,7,8,5,4,3,6,7,3]
y = [2,3,4,5,7,6,5,4,3,6,4]

plt.scatter(x,y,s=300,color='gold',marker='*')
plt.title("scatter plot")
plt.show()
```

```python
x = [3,4,6,7,8,5,4,3,6,7,3]
y = [2,3,4,5,7,6,5,4,3,6,4]
plt.bar(x,y,color='g',width=0.05)
plt.scatter(x,y,s=300,color='gold',marker="*",edgecolors='orange',linewidths=10)
plt.scatter(x,y,color='brown')
plt.title("scatter plot")
plt.show()
```

```python
#pie chart
slices = [8,3,10,5,4]
activities = ['sona','padhai','phone','khana','games']
cols = ['r','g','gold','c','m']

plt.pie(slices,labels=activities,colors=cols,startangle=90,shadow=True,explode=(0,0.5,0,0.1,0),autopct='%1.2f%%')
plt.title('pie chart')
plt.show()
```
