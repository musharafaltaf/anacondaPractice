# nested for loop : a nested loop is a loop inside a loop 
### the "innner loop" will be executed one time for each iteration of the "outer loop"

```python
for i in range (1,4):
    for j in range (1,5):
        print(i,j)
```

```python
row = int(input("Enter a number of rows:"))
for i in range (row,0,-1):
    print(' * ' *i)
```

```python
row = int(input("Enter a number of rows:"))
for i in range (1,row):
    print(' * ' *i)
for i in range (row,0,-1):
    print(' * ' *i)
```

```python
row = int(input("Enter a number of rows:"))
for i in range (1,row):
    print(' * ' *i)
```

# while loop 
## python while loop is used to run a block code while a certain condition is met.
### the syntax of while loop is :
### initialization 
### while condition 
###  #body of while 
###  increment/decrement

```python
i = 1
while i<=5:
    print(i)
    i+=1
```
